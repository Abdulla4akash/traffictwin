## Load only the public TrafficTwin harness

This clone supplies the synthetic driver only. Do not add credentials, private remotes, or external assets.

```python
import shutil
import subprocess
import sys
from pathlib import Path

REPOSITORY_URL = "https://github.com/Abdulla4akash/traffictwin.git"
BRANCH = "claude/complete-v0.7"
REPOSITORY_DIR = Path("/content/traffictwin-bbus-smoke")
if REPOSITORY_DIR.exists():
    shutil.rmtree(REPOSITORY_DIR)
subprocess.run(
    ["git", "clone", "--depth", "1", "--branch", BRANCH, REPOSITORY_URL, str(REPOSITORY_DIR)],
    check=True,
)
sys.path.insert(0, str(REPOSITORY_DIR))
commit = subprocess.run(
    ["git", "rev-parse", "HEAD"], cwd=REPOSITORY_DIR, check=True, capture_output=True, text=True
).stdout.strip()
print({"branch": BRANCH, "commit": commit})
```

## Run the declared synthetic matrix

The fixed default uses five model seeds per training fixture and common held-out evaluation seeds. Do not tune it from the outputs.

```python
from gpu.colab.bbus_synthetic_trace_smoke import TraceSmokeConfig, run_experiment

OUTPUT_DIR = Path("/content/bbus-synthetic-trace-smoke-output")
if OUTPUT_DIR.exists():
    shutil.rmtree(OUTPUT_DIR)
config = TraceSmokeConfig(
    model_seeds=(200, 201, 202, 203, 204),
    evaluation_seeds=(9100, 9101, 9102),
    updates=250,
    batch_size=2048,
    training_trace_size=32768,
    evaluation_trace_size=16384,
    require_gpu=True,
)
result = run_experiment(config, OUTPUT_DIR)
print(json.dumps({
    "output_dir": result["output_dir"],
    "design_fingerprint": result["design_fingerprint"],
    "scientific_evidence": False,
}, indent=2, sort_keys=True))
```
